# 1. Top 10 Restaurants by Weighted Rating

```python
restaurant_score = (
    df.groupby("Restaurant Name")
      .agg(
          avg_rating=("Rating", "mean"),
          total_reviews=("Rating Count", "sum")
      )
)

restaurant_score["weighted_score"] = (
    restaurant_score["avg_rating"] *
    np.log1p(restaurant_score["total_reviews"])
)

restaurant_score.sort_values(
    "weighted_score",
    ascending=False
).head(10)
```

```python
Insight:Finds restaurants that are both highly rated and highly reviewed.
```

# 2. Most Popular Dishes

```python
popular_dishes = (
    df.groupby("Dish Name")["Rating Count"]
      .sum()
      .sort_values(ascending=False)
      .head(20)
)

print(popular_dishes)
```

```python
Insight:Identifies dishes receiving the most customer engagement.
```

# 3. Premium Dishes with High Ratings

```python
premium = (
    df.groupby("Dish Name")
      .agg(
          avg_price=("Price (INR)", "mean"),
          avg_rating=("Rating", "mean")
      )
      .query("avg_price > 300")
      .sort_values("avg_rating", ascending=False)
)

print(premium.head(20))
```

```python
Insight:Shows expensive dishes that customers genuinely love.
```

# 4. Monthly Revenue Trend

```python
monthly_revenue = (
    df.groupby(df["Order Date"].dt.to_period("M"))
      ["Price (INR)"]
      .sum()
)

print(monthly_revenue)
```

```python
Insight:Tracks revenue growth and seasonality.
```

# 8. Top Restaurant in Every City

```python
city_top = (
    df.groupby(["City","Restaurant Name"])
      .agg(avg_rating=("Rating","mean"))
      .reset_index()
      .sort_values(
          ["City","avg_rating"],
          ascending=[True,False]
      )
      .drop_duplicates("City")
)

print(city_top)
```

# 9. Category Performance Dashboard

```python
category_perf = (
    df.groupby("Category")
      .agg(
          avg_price=("Price (INR)", "mean"),
          avg_rating=("Rating", "mean"),
          total_reviews=("Rating Count", "sum")
      )
      .sort_values(
          "total_reviews",
          ascending=False
      )
)

print(category_perf)
```

```python
Insight:Compares categories by price, quality, and popularity.
```

# 10. Hidden Gems Analysis

```python
Highly rated dishes with low review counts
```

```python
hidden_gems = (
    df.groupby("Dish Name")
      .agg(
          avg_rating=("Rating","mean"),
          reviews=("Rating Count","sum")
      )
      .query("avg_rating >= 4.7 and reviews < 500")
      .sort_values("avg_rating", ascending=False)
)

print(hidden_gems.head(20))
```

```python
Insight:Finds underrated dishes that customers love but few people have discovered.
```

# 11. Top 20 Dishes by Popularity Score

```python
dish_score = (
    df.groupby("Dish Name")
      .agg(
          avg_rating=("Rating","mean"),
          reviews=("Rating Count","sum")
      )
)

dish_score["Popularity Score"] = (
    dish_score["avg_rating"] *
    np.log1p(dish_score["reviews"])
)

top_dishes = dish_score.sort_values(
    "Popularity Score",
    ascending=False
).head(20)

print(top_dishes)
```

```python
Insight:Ranks dishes using both customer satisfaction and review volume, producing a more reliable popularity metric than rating alone.
```

# 12. Customer Demand Hotspots (City × Category Analysis)
# Question : Which food categories are most popular in each city?

```python
city_category = (
    df.pivot_table(
        values="Rating Count",
        index="City",
        columns="Category",
        aggfunc="sum",
        fill_value=0
    )
)

print(city_category)
```

```python
Insight : Identifies regional food preferences and helps restaurants optimize menus by city.
```

# 13. Price Segmentation Analysis
# Question:How do ratings differ across price segments?

```python
df["Price Segment"] = pd.cut(
    df["Price (INR)"],
    bins=[0,100,200,300,500,1000],
    labels=[
        "Budget",
        "Economy",
        "Standard",
        "Premium",
        "Luxury"
    ]
)

segment_analysis = (
    df.groupby("Price Segment")
      .agg(
          avg_price=("Price (INR)", "mean"),
          avg_rating=("Rating", "mean"),
          total_reviews=("Rating Count", "sum")
      )
)

print(segment_analysis)
```

```python
Insight:Shows whether premium-priced dishes actually deliver higher customer satisfaction.
```

# 14. Restaurant Market Share Analysis
# Question : Which restaurants dominate customer attention?

```python
total_reviews = df["Rating Count"].sum()

market_share = (
    df.groupby("Restaurant Name")["Rating Count"]
      .sum()
      .reset_index()
)

market_share["Market Share (%)"] = (
    market_share["Rating Count"] /
    total_reviews * 100
)

market_share = market_share.sort_values(
    "Market Share (%)",
    ascending=False
)

print(market_share.head(20))
```

```python
Insight:Measures restaurant dominance based on customer engagement and visibility.
```

# 15. Best Value-for-Money Dishes
# Question:Which dishes offer the highest rating relative to price?

```python
value_dishes = (
    df.groupby("Dish Name")
      .agg(
          avg_rating=("Rating", "mean"),
          avg_price=("Price (INR)", "mean"),
          reviews=("Rating Count", "sum")
      )
)

value_dishes = value_dishes.query("reviews > 500")

value_dishes["Value Score"] = (
    value_dishes["avg_rating"] /
    value_dishes["avg_price"]
)

value_dishes = value_dishes.sort_values(
    "Value Score",
    ascending=False
)

print(value_dishes.head(20))
```

```python
Insight:Identifies dishes delivering the highest customer satisfaction per rupee spent.
```
